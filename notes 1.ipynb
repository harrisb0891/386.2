```python
# single dimensional arrays (Vector)

# Array type are determined automatically by numpy
# Declare a numpy array with list, tuple, range (sequence)

oneD = np.array(
    [1, 2, 3, 4, 5, 6]
)

print(oneD)
print(oneD[1:4:2])
```

### Declaring Arrays in Python
-   `.array()` - delarcing simple arrays with lists, tuples, ranges
-   `.arange()` - rangle function to create an array.
    -   syntax: `np.arange(start, stop, step, dtype)`

```python
one_to_100 = np.arange(100)

print(one_to_100)

# array from 1 - 30 not including 30
one_thirty = np.arange(1, 30)
print(one_thirty)

# array from 10-20 by 2
ten_twenty = np.arange(10, 20, 2)
print(ten_twenty)

# array from -3 -10 10 by 4, verify dtype is int
print(np.arange(-3, 10, 4, dtype=int))
```

#### NumPy Array Attributes - Array descriptors
-   `array.shape` - tuple with the num of rows and columns
-   `array.size` - total number of elements (records) in array
-   `array.dtype` - data type of arrays
-   `array.ndim` - number of dimensions

```python
print(oneD)
print(oneD.shape)
print(oneD.size)
print(oneD.dtype)
print(oneD.ndim)
```

### NumPy Operations

```python
# Array Access
arr = np.array([1, 2, 3, 4, 5])

element = arr[3]
print('Element: ', element)

slice = arr[1:4]
print('Slice: ', slice)

print(arr)
```

```python
# Arithmetic and NumPy Arrays
# Arithmetic with arrays and scalar values compute element-wise (PER ELEMENT, for each element)
arr = np.array([1, 2, 3])
arr2 = np.array([9, 8, 7])

print(arr + 1) #[2, 3, 4]
print(arr * 3) #[3, 6, 9]
print(arr / 2) # [0.5, 1.0, 1.5]

print(arr + arr2) # [10, 10, 10] added elements at the same index together.
```

```python
# NumPy Aggregations
arr = np.array([1, 2, 3, 4, 5])

sum = np.sum(arr)
mean = np.mean(arr)
median = np.median(arr)
stDev = np.std(arr)
min = np.min(arr)
max = np.max(arr)

print('Sum: ', sum)
print('Avg: ', mean)
print('Median: ', median)
print('Standard Deviation: ', stDev)
print('Min: ', min)
print('Max: ', max)
```

```python
# Transpose or Reshape
matrix = np.array(
    [
        [1, 2, 3],
        [4, 5, 6]
    ]
)

print('Before Transpose')
print(matrix)
# Tranpose, pivot table 90 degrees
transpose_matrix = matrix.T
print('\nTransposed Matrix')
print(transpose_matrix)


reshaped_matrix = matrix.reshape(3, 2)

print('\nReshaped Matrix')
print(reshaped_matrix)
```

```python
# Logical Operators AND, OR NOT
a = np.array([True, True, False, False])
b = np.array([True, False, True, False])
print('Logical And') # compare each element in a list
print(np.logical_and(a, b))

print('\nLogical Or') # compare each element in a list
print(np.logical_or(a, b))
```

```python
c = np.array([True, True, False, False])
print(np.logical_not(c)) # Flips the boolean value
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([1, 2, 3, 4])
c = np.array([2, 3, 4, 5])

print(np.array_equal(a, b)) # compares the entire array looking for differences.

print(np.array_equal(a, c))
```

```python
# Sorting Arrays
a = np.array([3, 13, 6, 23, 14, 100, -1, -8, 32])

print(np.sort(a))
```

```python
# Searching

# Membership operators
print(12 not in a)
print(3 in a)
```

```python
b = np.array([3, 13, 6, 23, 14, 100, -1, -8, 32])
print(np.argmin(b)) # return the index number of the lower/minimum value

print(np.argmax(b)) # return the index number of the max value
```

```python
# count_nonzero() - counts non-zero elements
a = np.array([0, 1, 2, 0, 13, 15, 0, 6])

print(np.count_nonzero(a))

b = np.array([
    [1, 2, 0],
    [0, 0, 0]
])

print(np.count_nonzero(b))
```

```python
# concatenate() - combines multiple arrays into one array

a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])
c = np.concatenate((a, b), axis=0)

print(c)
```
